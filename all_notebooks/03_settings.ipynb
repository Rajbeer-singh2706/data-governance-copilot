# 03 — Configuration (`config/settings.py`)
`AppConfig` is a dataclass built from environment variables at import time.  
It contains nested configs: `LLMConfig`, `DatabricksConfig`, `RedisConfig`, `VectorDBConfig`.  
A singleton `config = AppConfig()` is used everywhere else.


```
import sys, os
sys.path.insert(0, os.path.join(os.getcwd(), '..', 'codebase', 'codebase', 'src'))
os.makedirs("logs", exist_ok=True)
```

## 1. LLMConfig

```
from config.settings import LLMConfig
llm = LLMConfig()
print("provider      :", llm.provider)
print("primary_model :", llm.primary_model)
print("model (alias) :", llm.model)
print("temperature   :", llm.temperature)
print("max_tokens    :", llm.max_tokens)
print("timeout       :", llm.timeout)
print("max_retries   :", llm.max_retries)
print("fallback_models:", llm.fallback_models)
print("api_key set   :", bool(llm.api_key))
```

## 2. RedisConfig & URL property

```
from config.settings import RedisConfig
r = RedisConfig()
print("host    :", r.host)
print("port    :", r.port)
print("db      :", r.db)
print("enabled :", r.enabled)
print("url     :", r.url)
```

## 3. DatabricksConfig

```
from config.settings import DatabricksConfig
db = DatabricksConfig()
print("host configured :", bool(db.host))
print("catalog :", db.catalog)
print("schema  :", db.schema)
```

## 4. VectorDBConfig & connection_string

```
from config.settings import VectorDBConfig
v = VectorDBConfig()
print("host          :", v.host)
print("port          :", v.port)
print("database      :", v.database)
print("table_name    :", v.table_name)
print("embedding_dim :", v.embedding_dim)
print("conn_string   :", v.connection_string)
```

## 5. AppConfig — the singleton

```
from config.settings import AppConfig, config
print("environment :", config.environment)
print("debug       :", config.debug)
print("log_level   :", config.log_level)
print("LLM provider:", config.llm.provider)
print("Redis URL   :", config.redis.url)
print("Databricks  :", bool(config.databricks.host))
```

## 6. DATA_PRODUCTS registry

```
from config.settings import DATA_PRODUCTS
for product, info in DATA_PRODUCTS.items():
    print(f"\n{product.upper()}")
    print(f"  description : {info['description']}")
    print(f"  owner       : {info['owner']}")
    print(f"  table       : {info['table']}")
    print(f"  key_metrics : {info['key_metrics']}") 
```

## 7. Override config via environment variables

```
import os
os.environ["LLM_PROVIDER"] = "openai"
os.environ["LLM_MODEL"] = "gpt-4o-mini"

# Build a fresh config (the module singleton doesn't reload)
fresh = AppConfig()
print("provider :", fresh.llm.provider)
print("model    :", fresh.llm.primary_model)

# Reset
os.environ["LLM_PROVIDER"] = "groq"
os.environ["LLM_MODEL"] = "llama-3.3-70b-versatile" 
```
