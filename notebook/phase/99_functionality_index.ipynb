# Functionality Index

This project currently exposes these implemented functionality areas and matching notebooks.

| Functionality | What It Covers | Notebook |
|---|---|---|
| Complete setup diagnostics | Python, package versions, config, imports, Redis, Docker status | `00_complete_setup_diagnostics.ipynb` |
| Redis cache | Redis connection, cache helpers, TTL, `@cached_node`, app keys | `02_complete_redis_functionality.ipynb` |
| Specialist agents | Information, Knowledge, Metadata, Capacity/Jira, Rule registry | `03_agents_functionality.ipynb` |
| Guardrails, intent, routing | PII redaction, blocking rules, intent fallback, route map | `04_guardrails_intent_routing.ipynb` |
| LangGraph orchestration and memory | Initial state, graph compile, checkpointer, optional graph query | `05_graph_orchestration_memory.ipynb` |
| LLM, logging, retry, base contracts | LLM factory config, agent dataclasses, logging, retry decorator | `06_llm_logging_retry.ipynb` |
| Streamlit UI smoke | UI file, launch command, safe source checks, session-state expectations | `07_streamlit_ui_smoke.ipynb` |
| Docker/runtime | Compose config, running containers, app/Redis env, Redis keys | `08_docker_runtime_diagnostics.ipynb` |

Notes:

- API and Teams functionality is described in project docs, but matching `src/api` and `src/teams` modules are not present in the current working tree, so no executable notebook was created for those surfaces.
- `01_redis_cache.ipynb` already existed and appears to be the earlier Redis learning notebook.
- Optional cells that can call an LLM or invoke the full graph are disabled by default where appropriate.
