## Running without credentials

Every notebook starts with:
```python
os.environ['ENABLE_MOCK'] = 'true'
```
This means **all services use mocks** — no Databricks, Jira, Collibra, pgvector, Redis,  
or LLM API keys are needed.

## Enabling real services

Copy `.env.example` → `.env`, set your credentials, then remove the `ENABLE_MOCK` line  
from the notebook cell before running.

| Credential | Purpose |
|------------|--------|
| `GROQ_API_KEY` | LLM synthesis (notebooks 08, 14, 18) |
| `DATABRICKS_HOST` + `DATABRICKS_TOKEN` | Real metrics (notebook 02, 13) |
| `JIRA_URL` + `JIRA_TOKEN` | Real Jira tickets (notebook 05, 12) |
| `COLLIBRA_URL` + credentials | Real metadata (notebook 04, 13) |
| `DATABASE_URL` | pgvector + Postgres checkpointer (notebooks 03, 22) |
| `REDIS_HOST` | Redis cache + rate limiting (notebooks 10, 14) |
