## Try the full classifier (uses MockLLM if no API key)

```
from graph.intent import classify_intent
from config.settings import get_config
cfg = get_config()

test_queries = [
    'Why did our gross retention rate drop in Q1?',
    'Create a data quality rule for null checks on bookings',
    'Who is the steward of the LTV dataset?',
]
for q in test_queries:
    try:
        result = classify_intent(q, cfg)
        print(f'\nQuery: {q}')
        print(f'  Intent: {result.intent.value}  Confidence: {result.confidence:.2f}')
        print(f'  Products: {result.data_products}')
        print(f'  Reasoning: {result.reasoning}')
    except Exception as e:
        print(f'Classifier error (expected without API key): {e}')
```
