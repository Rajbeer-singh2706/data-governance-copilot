## Override with environment variables
Set env vars before calling get_config() to customise behaviour.

```python
from config.settings import reset_config
os.environ['LLM_MODEL'] = 'gpt-4o-mini'
os.environ['ENVIRONMENT'] = 'staging'
reset_config()
cfg2 = get_config()
print('Overridden model:', cfg2.llm.model)
print('Overridden env:', cfg2.environment)
# Restore
del os.environ['LLM_MODEL']; del os.environ['ENVIRONMENT']
reset_config()
```
