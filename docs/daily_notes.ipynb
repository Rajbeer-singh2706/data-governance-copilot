**Title: Build an Agentic AI Data Governance & Analytics Assistant**
**Objective:**
Design and implement an Agentic AI system that acts as a conversational interface (via Microsoft Teams or Web UI) for business users to interact with enterprise data products (e.g., Bookings, Retention, LTV, CAC). The system should intelligently orchestrate multiple specialized agents to retrieve, analyze, and act on data-related queries.
**System Overview:**
The system must include a **Supervisor Agent (Orchestrator)** that interprets user queries, determines intent, and dynamically routes requests to relevant specialized agents. It should aggregate responses and present a concise, business-friendly summary.
**Core Capabilities:**
### 1. Read (Insight Generation)
The Supervisor Agent should collaborate with the following agents:
* **Knowledge Agent**
  * Parse and retrieve insights from unstructured sources (PDFs, PPTs, DOCX, Excel)
  * Integrate with SharePoint / Confluence
  * Provide business definitions, documentation, and contextual knowledge
* **Information Agent**
  * Query structured data sources (Databricks, Data Warehouse, SQL)
  * Provide metrics, facts, dimensions, and data lineage
  * Support data products like Bookings, Retention, LTV, CAC
* **Metadata Agent (Read via MCP)**
  * Integrate with Data Governance tools (e.g., Collibra)
  * Fetch business metadata, classifications, ownership (stewardship)
  * Retrieve data quality scores and governance attributes
* **Capacity Agent (Read)**
  * Integrate with Jira or similar tools
  * Fetch stories, bugs, and incidents related to data products
  * Provide operational context (ongoing issues, delays, failures)
### 2. Write (Action Execution)
The system should also support action-taking capabilities:
* **Rule Agent**
  * Create and manage business rules and data quality rules
* **Metadata Agent (Write)**
  * Update business metadata, classifications, and governance attributes
* **Capacity Agent (Write)**
  * Create Jira tickets (bugs, stories, tasks) based on detected issues
**Supervisor Agent Responsibilities:**
* Parse natural language queries from users
* Identify intent (metrics, documentation, data quality, issues, etc.)
* Route requests to one or multiple agents
* Aggregate and reconcile responses
* Generate a unified, concise, and business-friendly summary
* Trigger write actions when required
**Example Query:**
“Why did retention drop last month?”
**Expected Behavior:**
* Information Agent → Fetch retention metrics
* Metadata Agent → Retrieve data quality scores
* Capacity Agent → Identify related Jira issues
* Knowledge Agent → Provide business context
* Supervisor Agent → Combine all insights into a final explanation
**Tech Stack (Suggested):**
* Orchestration: LangGraph / CrewAI
* LLM: OpenAI / Azure OpenAI
* RAG: FAISS / Chroma (for Knowledge Agent)
* Data Sources: Databricks / SQL Warehouse
* Governance: Collibra (via MCP)
* Ticketing: Jira API
* UI: Streamlit / Teams Bot
**Expected Output:**
A scalable, modular, multi-agent system capable of:
* Answering business questions
* Explaining data issues
* Providing governance insights
* Triggering actions across systems
**Constraints:**
* Modular agent design (each agent independently extensible)
* Clear separation of read vs write responsibilities
* Production-ready structure (logging, error handling, configs)
* Secure API integrations
**Goal:**
Build a real-world **Data Governance Copilot** that enhances decision-making, reduces manual effort, and bridges the gap between business users and complex data ecosystems.
