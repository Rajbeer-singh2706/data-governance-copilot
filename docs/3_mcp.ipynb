### 🟢 New: MCP Server (`src/mcp_server/server.py`)

**10 tools exposed:**

| Tool | What it does |
|---|---|
| `governance_query` | Full multi-agent pipeline — intent → agents → synthesis |
| `query_metrics` | Information Agent (Databricks SQL / metrics) |
| `search_knowledge_base` | Knowledge Agent (pgvector RAG search) |
| `get_metadata` | Metadata Agent (Collibra data quality) |
| `manage_incidents` | Capacity Agent (Jira tickets / SLA) |
| `manage_rules` | Rule Agent (CRUD governance rules) |
| `get_system_status` | Redis health, agent status, token usage |
| `invalidate_cache` | Flush Redis/in-memory cache by agent or all |
| `approve_action` | HITL approval for pending Jira ticket creation |
| `ingest_document_url` | Download URL → chunk → embed → upsert to Neon |

**Two transports:**
- `stdio` (default) — for Claude Desktop, zero config
- `SSE` (`TRANSPORT=sse`) — for remote/web clients, listens on port 8002, also added to `compose.yml`
